**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
# Write your code here for Q7
class Animal:
    def __init__(self):
        self.living = True
    def eat(self):
        print("animal can eat")
class Mammal(Animal):
    def __init__(self):
        super().__init__()
        self.warm_blooded = True
    def walk(self):
        print("Mmammal can walk")

class Cat(Mammal):
    def __init__(self):
        super().__init__()
        self.breed = "parsian"
    def voice(self):
        print("bakra bole mai mai mai")
```

**Q8:** Multiple Inheritance ka istemal karte hue do parent classes `Father` aur `Mother` banayein, aur unse aik child class `Child` derive karein jo dono ke habits inherit kare.

```python
class Father:
    def __init__(self):
        self.father_habit = "Hard work"
    def father_skill(self):
        print("father skill: rosting")

class Mother:
    def __init__(self):
        self.mother_habit = "Caring"
    def mother_skill(self):
        print("mother skill: rosting")

class Child(Father, Mother):
    def __init__(self):
        Father.__init__(self)
        Mother.__init__(self)
        self.child_habit = "gaming"
        
c = Child()
print(f"Father_habit: {c.father_habit}")
print(f"mother_ habit: {c.mother_habit}")
print(f"Child abit: {c.child_habit}")
c.father_skill()
c.mother_skill()    
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10
class Shape:
    def info(self):
        print("its a shape")

class Circle(Shape):
    def area(self, r):
        return 3.14 * r * r

class Square(Shape):
    def area(self, side):
        return side * side

c = Circle()
s = Square()
c.info()
print("Circle Area:", c.area(55))
print("Square Area:", s.area(40))
```

"**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

p1 = Point(234, 567)
p2 = Point(200,700)
p3 = p1 + p2
print(p3)
```

**Q19:** Abstraction aur Encapsulation ke darmiyan kya wazeh farq (main difference) hai? Dono ke concepts ko short statements mei define karein.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCard(Payment):
    def pay(self, amount):
        print(f"{amount} paid using Credit Card")

class jazzcase(Payment):
    def pay(self, amount):
        print(f"{amount} paid using jazzcase")

class Wallet:
    def __init__(self):
        self.__balance = 100000

    def get_balance(self):
        return self.__balance

    def add_money(self, amount):
        self.__balance += amount
        print(f"{amount} added")

print("--- Abstracration Testing ---")
p1 = CreditCard()
p1.pay(50000)
      
p2 = jazzcase()
p2.pay(20000)

print("\n=== Encapsulation Testing ---")
w = Wallet()
print(f"Current Balance: {w.get_balance()}")  
w.add_money(5000)
print(f"New Balance: {w.get_balance()}")
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

    def payment_successful(self, amount):
        print(f"Payment of {amount} Rs succesful")

class CreditCardPayment(Payment):        
    def __init__(self, card_number):
        self.card_number = card_number

    def pay(self, amount):
        print(f"Paying {amount} Rs using Credit Card ending with {self.card_number[-4:]}")

class UPIPayment(Payment):
    def __init__(self, upi_id):
        self.upi_id = upi_id
         
    def pay(self, amount):
        print(f"{amount} paid using UPI ID {self.upi_id}")

class Wallet:
    def __init__(self):
        self.__balance = 100

    def get_balance(self):
        return self.__balance
    
credit = CreditCardPayment("4321-8765-0921-6543")
credit.pay(5100)
upi = UPIPayment("abdullah@hunyar")
upi.pay(2100)

w = Wallet()
print(f"Wallet Balance: {w.get_balance()}")
       
```
